```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('insurance_pre.csv')
```

```python
dataset = pd.get_dummies(dataset, dtype=int, drop_first=True)
dataset
```

```python
independent = dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
dependent = dataset[['charges']]
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'criterion':['squared_error','absolute_error','poisson'],
'max_features': ['auto','sqrt','log2'],
'n_estimators':[10,100,1000,2000,3000]}
grid = GridSearchCV(RandomForestRegressor(), param_grid, refit=True, verbose=3, n_jobs=-1)
grid.fit(independent, dependent)
```

```python
final = grid.cv_results_
grid_predict = grid.predict(independent)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(grid_predict,dependent)
print('The R_Score value for the best parameter {}'.format(grid.best_params_),r_score)
```

```python
table = pd.DataFrame.from_dict(final)
table
```

```python
age=int(input('Enter the age:'))
bmi=float(input('Enter the bmi:'))
child=int(input('Enter the children:'))
gender=int(input('Enter the gender male(0) or female(1):'))
smoker=int(input('Enter the smoker yes(1) or no(0):'))
Future_prediction = grid.predict([[age,bmi,child,gender,smoker]])
print("insurance prediction = {}".format(Future_prediction))
```
